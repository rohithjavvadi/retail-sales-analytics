```python
# KPI 29
pd.read_sql("""
SELECT
    COUNT(DISTINCT CASE
        WHEN DATE(order_delivered_customer_date) >
             DATE(order_estimated_delivery_date)
        THEN order_id
    END) AS late_orders,

    COUNT(DISTINCT order_id) AS total_orders,

    ROUND(
        COUNT(DISTINCT CASE
            WHEN DATE(order_delivered_customer_date) >
                 DATE(order_estimated_delivery_date)
            THEN order_id
        END) * 100.0 /
        COUNT(DISTINCT order_id),
        2
    ) AS late_delivery_percentage
FROM orders
WHERE order_delivered_customer_date IS NOT NULL;
""", conn)
```

# Delivery KPI 29: Late Delivery Rate

### Business Question
What percentage of customer orders were delivered after the estimated delivery date?

### Objective
Measure the frequency of late deliveries.

### SQL Functions Used
- `CASE WHEN`
- `COUNT(DISTINCT)`
- `DATE()`
- `ROUND()`

### Why It Matters
Late deliveries negatively impact customer satisfaction, increase support requests, and may reduce customer retention.

### Insight
The query calculates the total number and percentage of orders delivered after the promised delivery date, providing a key indicator of logistics performance.

```python
# KPI 30
pd.read_sql("""
SELECT
    ROUND(
        AVG(
            JULIANDAY(order_delivered_customer_date) -
            JULIANDAY(order_estimated_delivery_date)
        ),
        2
    ) AS average_delivery_delay_days
FROM orders
WHERE order_delivered_customer_date IS NOT NULL
  AND order_estimated_delivery_date IS NOT NULL;
""", conn)
```
