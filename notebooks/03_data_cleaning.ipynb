```python
# ==========================================================
# Day 3: Data Cleaning & Preprocessing
# Project: RetailIQ Analytics Platform
# ==========================================================

# Import Libraries
import pandas as pd
import numpy as np
import os

# Display Settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", 120)
pd.set_option("display.max_colwidth", None)

print("=" * 60)
print("RetailIQ Analytics Platform")
print("Day 3 - Data Cleaning & Preprocessing")
print("Libraries Imported Successfully")
print("=" * 60)
```

```python
# ==========================================================
# Load Raw Datasets
# ==========================================================

customers = pd.read_csv("../data/raw/olist_customers_dataset.csv")
orders = pd.read_csv("../data/raw/olist_orders_dataset.csv")
order_items = pd.read_csv("../data/raw/olist_order_items_dataset.csv")
payments = pd.read_csv("../data/raw/olist_order_payments_dataset.csv")
reviews = pd.read_csv("../data/raw/olist_order_reviews_dataset.csv")
products = pd.read_csv("../data/raw/olist_products_dataset.csv")
sellers = pd.read_csv("../data/raw/olist_sellers_dataset.csv")
geolocation = pd.read_csv("../data/raw/olist_geolocation_dataset.csv")
category_translation = pd.read_csv("../data/raw/product_category_name_translation.csv")

print("All datasets loaded successfully!\n")

datasets = {
    "Customers": customers,
    "Orders": orders,
    "Order Items": order_items,
    "Payments": payments,
    "Reviews": reviews,
    "Products": products,
    "Sellers": sellers,
    "Geolocation": geolocation,
    "Category Translation": category_translation
}

for name, df in datasets.items():
    print(f"{name:<22} Shape: {df.shape}")
```

```python
# ==========================================================
# Create Working Copies
# ==========================================================

customers_clean = customers.copy()
orders_clean = orders.copy()
order_items_clean = order_items.copy()
payments_clean = payments.copy()
reviews_clean = reviews.copy()
products_clean = products.copy()
sellers_clean = sellers.copy()
geolocation_clean = geolocation.copy()

print("Working copies created successfully!")
```

```python
# ==========================================================
# Step 4: Dataset Inspection
# ==========================================================

def inspect_dataset(df, name):
    print("=" * 80)
    print(f"Dataset: {name}")
    print("=" * 80)

    # Shape
    print(f"\nShape: {df.shape}")

    # Data Types
    print("\nData Types:")
    print(df.dtypes)

    # Missing Values
    print("\nMissing Values:")
    missing = df.isnull().sum()
    missing = missing[missing > 0]

    if len(missing) == 0:
        print("No missing values.")
    else:
        print(missing.sort_values(ascending=False))

    # Duplicate Rows
    print(f"\nDuplicate Rows: {df.duplicated().sum()}")

    # Memory Usage
    memory = df.memory_usage(deep=True).sum() / 1024**2
    print(f"\nMemory Usage: {memory:.2f} MB")

    print("\nFirst 5 Rows:")
    display(df.head())

    print("\n")
```

```python
datasets = {
    "Customers": customers_clean,
    "Orders": orders_clean,
    "Order Items": order_items_clean,
    "Payments": payments_clean,
    "Reviews": reviews_clean,
    "Products": products_clean,
    "Sellers": sellers_clean,
    "Geolocation": geolocation_clean
}

for name, df in datasets.items():
    inspect_dataset(df, name)
```

```python
# ==========================================================
# Customers Cleaning
# ==========================================================

print("Customers Shape Before:", customers_clean.shape)

# Remove duplicate rows
customers_clean.drop_duplicates(inplace=True)

# Check missing values
print("\nMissing Values:")
print(customers_clean.isnull().sum())

print("\nCustomers Shape After:", customers_clean.shape)
```

```python
# ==========================================================
# Orders Cleaning
# ==========================================================

print("Orders Shape Before:", orders_clean.shape)

# Remove duplicate rows
orders_clean.drop_duplicates(inplace=True)

# Convert date columns to datetime
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders_clean[col] = pd.to_datetime(orders_clean[col])

print("\nMissing Values:")
print(orders_clean.isnull().sum())

print("\nOrders Shape After:", orders_clean.shape)
```

```python
print("Order Items Shape Before:", order_items_clean.shape)

order_items_clean.drop_duplicates(inplace=True)

print("\nMissing Values:")
print(order_items_clean.isnull().sum())

print("\nOrder Items Shape After:", order_items_clean.shape)
```

```python
print("Payments Shape Before:", payments_clean.shape)

payments_clean.drop_duplicates(inplace=True)

print("\nMissing Values:")
print(payments_clean.isnull().sum())

print("\nPayments Shape After:", payments_clean.shape)
```

```python
print("Reviews Shape Before:", reviews_clean.shape)

reviews_clean.drop_duplicates(inplace=True)

# Convert dates
reviews_clean["review_creation_date"] = pd.to_datetime(
    reviews_clean["review_creation_date"]
)

reviews_clean["review_answer_timestamp"] = pd.to_datetime(
    reviews_clean["review_answer_timestamp"]
)

print("\nMissing Values:")
print(reviews_clean.isnull().sum())

print("\nReviews Shape After:", reviews_clean.shape)
```

```python
print("Products Shape Before:", products_clean.shape)

products_clean.drop_duplicates(inplace=True)

print("\nMissing Values:")
print(products_clean.isnull().sum())

print("\nProducts Shape After:", products_clean.shape)
```

```python
print("Sellers Shape Before:", sellers_clean.shape)

sellers_clean.drop_duplicates(inplace=True)

print("\nMissing Values:")
print(sellers_clean.isnull().sum())

print("\nSellers Shape After:", sellers_clean.shape)
```

```python
print("Geolocation Shape Before:", geolocation_clean.shape)

geolocation_clean.drop_duplicates(inplace=True)

print("\nMissing Values:")
print(geolocation_clean.isnull().sum())

print("\nGeolocation Shape After:", geolocation_clean.shape)
```

```python
# ==========================================================
# Feature Engineering
# ==========================================================

orders_clean["purchase_year"] = (
    orders_clean["order_purchase_timestamp"].dt.year
)

orders_clean["purchase_month"] = (
    orders_clean["order_purchase_timestamp"].dt.month
)

orders_clean["purchase_day"] = (
    orders_clean["order_purchase_timestamp"].dt.day
)

orders_clean["purchase_weekday"] = (
    orders_clean["order_purchase_timestamp"].dt.day_name()
)

orders_clean["delivery_days"] = (
    orders_clean["order_delivered_customer_date"]
    - orders_clean["order_purchase_timestamp"]
).dt.days
```

```python
print("=" * 80)
print("Final Dataset Summary")
print("=" * 80)

for name, df in datasets.items():
    print(f"{name:<20} {df.shape}")
```

```python
# ==========================================================
# Save Processed Files
# ==========================================================

os.makedirs("../data/processed", exist_ok=True)

customers_clean.to_csv("../data/processed/customers_clean.csv", index=False)
orders_clean.to_csv("../data/processed/orders_clean.csv", index=False)
order_items_clean.to_csv("../data/processed/order_items_clean.csv", index=False)
payments_clean.to_csv("../data/processed/payments_clean.csv", index=False)
reviews_clean.to_csv("../data/processed/reviews_clean.csv", index=False)
products_clean.to_csv("../data/processed/products_clean.csv", index=False)
sellers_clean.to_csv("../data/processed/sellers_clean.csv", index=False)
geolocation_clean.to_csv("../data/processed/geolocation_clean.csv", index=False)

print("All cleaned datasets saved successfully!")
```
