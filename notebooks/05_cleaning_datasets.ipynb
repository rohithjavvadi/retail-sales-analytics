```python
# ============================================================
# Sellers Dataset Cleaning
# ============================================================

print("=" * 60)
print("SELLERS DATASET")
print("=" * 60)

# -------------------------
# Basic Information
# -------------------------
print(f"Shape: {sellers.shape}")

print("\nData Types:")
print(sellers.dtypes)

# -------------------------
# Missing Values
# -------------------------
print("\nMissing Values:")
print(sellers.isnull().sum())

# -------------------------
# Duplicate Checks
# -------------------------
print("\nDuplicate Rows:", sellers.duplicated().sum())
print("Duplicate seller_id:",
      sellers["seller_id"].duplicated().sum())

# -------------------------
# Blank String Check
# -------------------------
print("\nBlank Strings:")
print(
    sellers.select_dtypes(include="object").apply(
        lambda col: (col.str.strip() == "").sum()
    )
)

# -------------------------
# Standardize Text Columns
# -------------------------
text_columns = ["seller_city", "seller_state"]

for col in text_columns:
    sellers[col] = sellers[col].str.strip()

# -------------------------
# Validation After Cleaning
# -------------------------
print("\nRemaining Missing Values:")
print(sellers.isnull().sum())

print("\nFinal Shape:", sellers.shape)
# -------------------------
# Save Clean Dataset
# -------------------------
sellers.to_csv(
    os.path.join(PROCESSED_PATH, "sellers_clean.csv"),
    index=False
)

print("\n✅ Sellers dataset cleaned and saved successfully.")


```

```python
# ============================================================
# Category Translation Dataset Cleaning
# ============================================================

print("=" * 60)
print("CATEGORY TRANSLATION DATASET")
print("=" * 60)

# -------------------------
# Basic Information
# -------------------------
print(f"Shape: {category_translation.shape}")

print("\nData Types:")
print(category_translation.dtypes)

# -------------------------
# Missing Values
# -------------------------
print("\nMissing Values:")
print(category_translation.isnull().sum())

# -------------------------
# Duplicate Checks
# -------------------------
print("\nDuplicate Rows:", category_translation.duplicated().sum())

print(
    "Duplicate Portuguese Categories:",
    category_translation["product_category_name"].duplicated().sum()
)

print(
    "Duplicate English Categories:",
    category_translation["product_category_name_english"].duplicated().sum()
)

# -------------------------
# Blank String Check
# -------------------------
print("\nBlank Strings:")
print(
    category_translation.select_dtypes(include="object").apply(
        lambda col: (col.str.strip() == "").sum()
    )
)

# -------------------------
# Standardize Text Columns
# -------------------------
text_columns = category_translation.select_dtypes(include="object").columns

for col in text_columns:
    category_translation[col] = category_translation[col].str.strip()

# -------------------------
# Final Validation
# -------------------------
print("\nRemaining Missing Values:")
print(category_translation.isnull().sum())

print("\nFinal Shape:", category_translation.shape)

# -------------------------
# Save Clean Dataset
# -------------------------
category_translation.to_csv(
    os.path.join(PROCESSED_PATH, "category_translation_clean.csv"),
    index=False
)

print("\n✅ Category Translation dataset cleaned and saved successfully.")
```

```python
# ============================================================
# Order Items Dataset Cleaning
# ============================================================

print("=" * 60)
print("ORDER ITEMS DATASET")
print("=" * 60)

# -------------------------
# Basic Information
# -------------------------
print(f"Shape: {order_items.shape}")

print("\nData Types:")
print(order_items.dtypes)

# -------------------------
# Missing Values
# -------------------------
print("\nMissing Values:")
print(order_items.isnull().sum())

# -------------------------
# Duplicate Checks
# -------------------------
print("\nDuplicate Rows:", order_items.duplicated().sum())
print("Duplicate order_id + order_item_id:",
      order_items.duplicated(subset=["order_id", "order_item_id"]).sum())

# -------------------------
# Blank String Check
# -------------------------
print("\nBlank Strings:")
print(
    order_items.select_dtypes(include="object").apply(
        lambda col: (col.str.strip() == "").sum()
    )
)

# -------------------------
# Data Type Conversion
# -------------------------
order_items["shipping_limit_date"] = pd.to_datetime(
    order_items["shipping_limit_date"]
)

# -------------------------
# Standardize Text Columns
# -------------------------
text_columns = ["order_id", "product_id", "seller_id"]

for col in text_columns:
    order_items[col] = order_items[col].str.strip()

# -------------------------
# Final Validation
# -------------------------
print("\nRemaining Missing Values:")
print(order_items.isnull().sum())

print("\nFinal Shape:", order_items.shape)

# -------------------------
# Save
# -------------------------
order_items.to_csv(
    os.path.join(PROCESSED_PATH, "order_items_clean.csv"),
    index=False
)

print("\n✅ Order Items dataset cleaned and saved successfully.")
```

```python
# ============================================================
# Payments Dataset Cleaning
# ============================================================

print("=" * 60)
print("PAYMENTS DATASET")
print("=" * 60)

# -------------------------
# Basic Information
# -------------------------
print(f"Shape: {payments.shape}")

print("\nData Types:")
print(payments.dtypes)

# -------------------------
# Missing Values
# -------------------------
print("\nMissing Values:")
print(payments.isnull().sum())

# -------------------------
# Duplicate Checks
# -------------------------
print("\nDuplicate Rows:", payments.duplicated().sum())

# -------------------------
# Blank String Check
# -------------------------
print("\nBlank Strings:")
print(
    payments.select_dtypes(include="object").apply(
        lambda col: (col.str.strip() == "").sum()
    )
)

# -------------------------
# Standardize Text Columns
# -------------------------
text_columns = ["order_id", "payment_type"]

for col in text_columns:
    payments[col] = payments[col].str.strip()

# -------------------------
# Final Validation
# -------------------------
print("\nRemaining Missing Values:")
print(payments.isnull().sum())

print("\nFinal Shape:", payments.shape)

# -------------------------
# Save
# -------------------------
payments.to_csv(
    os.path.join(PROCESSED_PATH, "payments_clean.csv"),
    index=False
)

print("\n✅ Payments dataset cleaned and saved successfully.")
```

```python
# ============================================================
# Reviews Dataset Cleaning
# ============================================================

print("=" * 60)
print("REVIEWS DATASET")
print("=" * 60)

# -------------------------
# Basic Information
# -------------------------
print(f"Shape: {reviews.shape}")

print("\nData Types:")
print(reviews.dtypes)

# -------------------------
# Missing Values
# -------------------------
print("\nMissing Values:")
print(reviews.isnull().sum())

# -------------------------
# Duplicate Checks
# -------------------------
print("\nDuplicate Rows:", reviews.duplicated().sum())
print("Duplicate review_id:",
      reviews["review_id"].duplicated().sum())

# -------------------------
# Blank String Check
# -------------------------
print("\nBlank Strings:")
print(
    reviews.select_dtypes(include="object").apply(
        lambda col: (col.str.strip() == "").sum()
    )
)

# -------------------------
# Data Type Conversion
# -------------------------
date_columns = [
    "review_creation_date",
    "review_answer_timestamp"
]

for col in date_columns:
    reviews[col] = pd.to_datetime(reviews[col])

# -------------------------
# Standardize Text Columns
# -------------------------
text_columns = [
    "review_id",
    "order_id",
    "review_comment_title",
    "review_comment_message"
]

for col in text_columns:
    reviews[col] = reviews[col].fillna("").str.strip()

# -------------------------
# Final Validation
# -------------------------
print("\nRemaining Missing Values:")
print(reviews.isnull().sum())

print("\nFinal Shape:", reviews.shape)

# -------------------------
# Save
# -------------------------
reviews.to_csv(
    os.path.join(PROCESSED_PATH, "reviews_clean.csv"),
    index=False
)

print("\n✅ Reviews dataset cleaned and saved successfully.")
```
