```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 20)
```

```python
customers = pd.read_csv("/Users/rohithjavvadi/Downloads/archive (1)/olist_customers_dataset.csv")

geolocation = pd.read_csv("/Users/rohithjavvadi/Downloads/archive (1)/olist_geolocation_dataset.csv")

orders = pd.read_csv("/Users/rohithjavvadi/Downloads/archive (1)/olist_orders_dataset.csv")

order_items = pd.read_csv("/Users/rohithjavvadi/Downloads/archive (1)/olist_order_items_dataset.csv")

payments = pd.read_csv("/Users/rohithjavvadi/Downloads/archive (1)/olist_order_payments_dataset.csv")

reviews = pd.read_csv("/Users/rohithjavvadi/Downloads/archive (1)/olist_order_reviews_dataset.csv")

products = pd.read_csv("/Users/rohithjavvadi/Downloads/archive (1)/olist_products_dataset.csv")

sellers = pd.read_csv("/Users/rohithjavvadi/Downloads/archive (1)/olist_sellers_dataset.csv")

category_translation = pd.read_csv(
    "/Users/rohithjavvadi/Downloads/archive (1)/product_category_name_translation.csv"
)
```

```python
#
datasets = {
    "customers": customers,
    "geolocation": geolocation,
    "orders": orders,
    "order_items": order_items,
    "payments": payments,
    "reviews": reviews,
    "products": products,
    "sellers": sellers,
    "category_translation": category_translation
}
```

```python
for name, df in datasets.items():
    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    print(df.head())

    print("\nShape:")
    print(df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\n")
```

```python
for name, df in datasets.items():

    print("=" * 70)
    print(name.upper())

    print(df.dtypes)
    

    print("\n")
```

```python
for name, df in datasets.items():

    print("=" * 70)
    print(name.upper())

    missing = df.isnull().sum()

    missing = missing[missing > 0]

    print(missing.sort_values(ascending=False))

    print("\n")
```

```python
for name, df in datasets.items():

    duplicates = df.duplicated().sum()

    print(f"{name}: {duplicates}")
```

```python
for name, df in datasets.items():

    print("=" * 70)

    print(name.upper())

    print(df.describe())
```

```python
for name, df in datasets.items():

    print("=" * 70)

    print(name.upper())

    print(df.describe(include="object"))
```

```python
for name, df in datasets.items():
    print("=" * 70)
    print(f"TABLE: {name.upper()}")
    print("=" * 70)

    summary = pd.DataFrame({
        "Column": df.columns,
        "Data Type": df.dtypes.values,
        "Missing Values": df.isnull().sum().values,
        "Unique Values": df.nunique().values
    })

    display(summary)
```

```python
for name, df in datasets.items():
    print("=" * 70)
    print(name.upper())

    for column in df.columns:
        if df[column].is_unique:
            print(f"{column} → Candidate Primary Key")
```
